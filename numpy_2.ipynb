# Indexing and reshaping:

```python
import numpy as np
```

```python
a = np.array([100, 200, 300, 400, 500, 600, 700, 800, 900])
```

```python
# Postive Indexing
print(a[1])
print(a[5])
print(a[:])
print(a[0])
```

```python
# Negative indexing 
print(a[-1]) # Last element
print(a[-2]) # 2nd Last Element
```

```python
#Slicing 
print(a[1:4])
print(a[::]) # All elements
print(a[::2])
print(a[::-1]) # All elements from backside with stepsize -1
print(a[::-2]) # All elements from backside with stepsize -2
```

```python
# Boolean Masking
mask = a > 250 # adding Boolean check 
a[mask]
```

```python
# Integer(fancy) indexing:
a[[0, 2, 4]] # Getting Multiple values
```

```python
# where condition: np.where(condition, value_if_true, value_if_false)
np.where(a > 250, 10000, 0000 )
```

```python
np.where(a < 250, 5000, 0000)
```

```python
b = np.array([0000, 3000, 0000, 7000, 0000, 9000])
```

```python
b
```

```python
np.nonzero(b) # Only nonzero values will index will display
```

# Shape Manipulation

```python
#reshape
x = np.arange(12)
```

```python
x.reshape(4,3)
```

```python
x.reshape(4, 1) # ValueError: cannot reshape array of size 12 into shape (4,1)
```

```python
# expand_dims
c = np.array([1, 2, 3])
```

```python
c
```

```python
np.expand_dims(c, axis = 0).shape # row vector, axis = 0 row vector
```

```python
np.expand_dims(c, axis = 1).shape # column vector, axis = 1
```

```python
#Squeeze: np.squeeze() removes all axes of length 1 from an array’s shape.
d = np.zeros((1, 3, 1))
np.squeeze(d).shape
# d = np.zeros((1, 3, 1)) creates an array of shape (1, 3, 1).
# np.squeeze(d) drops the two size-1 axes (the first and the last), leaving only the axis of length 3.
# np.squeeze(d).shape returns (3,).
```

# Joining and Splitting

```python
# concatenate(Joining along an existing axis)
a1 = np.array([[10, 20], [30, 40]])
b1 = np.array([[5, 6]])
np.concatenate([a1, b1], axis=0)
```

```python
#stack: joins along a new axis
c1 = np.array([100, 200, 300])
d1 = np.array([400, 500, 600])
e1 = np.stack([c1, d1], axis=0)
```

```python
e1.shape
```

```python
#split
f1 = np.arange(6)
np.split(f1, 3) # splitting the arrays
```

# Quick notes
### ravel vs flatten: 
ravel returns a view when it can (faster, shares memory), while flatten always copies.
### concatenate vs stack: 
concatenate keeps the number of dimensions the same, while stack adds a new one.
### np.where(cond) vs np.nonzero: 
with a single argument, np.where(cond) is equivalent to cond.nonzero().
