# NumPy Vectorization and Broadcasting

```python
import numpy as np
```

### 1. Elementwise operations

```python
a = 100
b = 200

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)
print(a+2)
print(a*10)
print(a>15)
```

### Array Functions such as sqrt, exp, log and clip

```python
print(np.sqrt([1, 4, 9])) # Finding Square root of values
print(np.sqrt(100)) # Sqrt returns the decimal value
print(np.round(np.exp([0, 1, 2]), 2))
print(np.log([1, np.e, np.e**2]))
scores = np.array([-5, 40, 85, 110])
print(np.clip(scores, 0, 100))
```

### Broadcasting Rules:
#### NumPy compares shapes from right to left. Two dimensions are compatible when they are equal or one of them is 1. A missing leading dimension is treated as 1.

```python
prices = np.array([[100, 200, 300], [400, 500, 600]])
prices.shape
discounts = np.array([10, 20, 30])
print(prices-discounts)
```

```python
# (3,) is treated as (1, 3), so the same three discounts apply to each row.

# To apply one discount per row, use shape (2, 1) instead.
row_discounts = np.array([[1000], [2000]])
row_discounts.shape
print(prices-row_discounts)
```

# Axis and Keepdims:
### For a 2D array, 
### axis=0 reduces the rows and returns one result per column. 
### axis=1 reduces the columns and returns one result per row.

```python
a = np.array([[10, 20, 30], [40, 50, 60]])
print(a.sum()) # Total sum of
print(a.sum(axis = 0)) # axis=0 reduces the rows and returns one result per column. 
print(a.sum(axis = 1)) # axis=1 reduces the columns and returns one result per row.
# axis = 0  reduces the rows and column wise opearion and reduces the rows, 
# axis = 1  reduces the column and row wise opearion and reduces the columns, 
```

```python
# keepdims = True preserves the reduced axis with size 1. That makes it easy to broadcast the result back across the original array.
row_means = a.mean(axis = 1, keepdims = True)
print(row_means)
print(a - row_means)
```

# Conditional calculations with np.where
### np.where(condition, value_if_true, value_if_false) selects a value for each element. Here, sales of at least ₹1,000 earn a 10% bonus.

```python
sales  = np.array([1000, 1000, 1500, 600])
bonus = np.where(sales >= 1000,  sales * 0.10, 0)
print(bonus)
print(sales[sales >= 1000])
```

# Pairwise operations with broadcasting
### To compare every element in x with every element in y, reshape x into a column using [:, None].

```python
x = np.array([10, 20, 30])
y = np.array([1, 2])

print(x.shape)
print(y.shape)
differences = x[1, None] - y
print(differences)
# Each row corresponds to one value from x, and each column corresponds to one value from y. For example, differences[1, 0] is 20 - 1 = 19.
```
