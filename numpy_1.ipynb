# NumPy Array Fundamentals With Examples:

```python
import numpy as np
```

```python
a = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
], dtype=np.int32)
```

```python
a
```

```python
# ndarray is NumPy’s main array type. An array can have multiple dimensions and has one dtype.
isinstance(a, np.ndarray) 
```

```python
#A scalar array is 0D, a vector is 1D, and a table of rows and columns is 2D.
np.array(7).ndim, np.array([1, 2, 3]).ndim, a.ndim
```

```python
# The length of each axis. Here, there are 3 rows and 3 columns.
print(a.shape)
```

```python
# The number of axes. It equals len(a.shape).
print(a.ndim)
print(len(a.shape))
```

```python
# The total number of elements. Here, 3 × 3 = 9.
a.size
```

```python
# The element data type. int32 is a signed 32-bit integer.
a.dtype
```

```python
# The number of bytes used by one element.
a.itemsize
```

```python
# The bytes counted for the array’s elements. Here, 6 × 4 = 24. Python object overhead is tracked separately.
a.nbytes
```

# Creating Arrays

```python
#1 Create an array from known values. Use nested lists for a 2D array.
numbers = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90])
print(numbers)
print(numbers[0])
print(numbers * 10)
```

```python
#2 np.asarray() converts a Python list or tuple into a NumPy array.
numbers = [100, 200, 300, 400, 500, 600, 700, 800, 900]
arr = np.asarray(numbers)
print(numbers)
print(arr * 2)

```

```python
same_arr = np.asarray(arr)
print(same_arr is arr)  # True
```

```python
#3  np.zeros() creates a NumPy array filled with zeros.
arr = np.zeros(10)
print(arr)
```

```python
# Create a 2D array with 2 rows and 3 columns
matrix = np.zeros((3, 3), dtype=int)
print(matrix)
```

```python
#4 np.ones Created a numpy array filled with ones.
```

```python
matrix = np.ones((4, 4), dtype=int)
print(matrix)
```

```python
matrix = np.ones((3, 3), dtype=float)
print(matrix)
```

```python
# dtype = int => 0
# dtype = float => 0.
```

```python
#5 np.full() creates a NumPy array and fills every element with the value you choose.
# In np.full(shape, fill_value), the first argument sets the array’s shape and the second sets the value to fill it with.
arr = np.full(5, 7)
print(arr)
```

```python
arr1 = np.full((3, 3), 9)
print(arr1)
```

```python
#6 np.arange() creates a NumPy array of evenly spaced numbers. In np.arange(start, stop, step), the stop value is excluded.
print(np.arange(10))
print(np.arange(10) * 2)
print(np.arange(3, 10)) # includes starting value and excludes ending value
print(np.arange(10, 20, 2)) # 2 indicate the step size.
print(np.arange(30, 10, -3)) # Staring point 30, and ending point 10 and step size -3


```

```python
#7 np.linspace() Generate a fixed number of evenly spaced values. Both endpoints are included by default.
print(np.linspace(10, 20, 5)) # 10 starting point, 20 ending point and 5 equally spacing 
```

```python
print(np.linspace(0, 10, num=5, endpoint=False)) # Generate 5 evenly spaced values from 0 up to 10, excluding 10
```

### With np.linspace(), you specify how many values you want. With np.arange(), you specify the step size.

```python
#8 np.eye() creates a 2D NumPy array with ones on the main diagonal and zeros everywhere else.
matrix = np.eye(3, dtype=int)
print(matrix)
# 3 gives you three rows and three columns. dtype=int gives you integer values.
```
